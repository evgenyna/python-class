## הערות בקוד

### הערת שורה: `#`

אפשר לכתוב טקסט חופשי, כמו הסבר או הערה, אם מתחילים את השורה ב־`#`. המחשב מתעלם מהשורה הזו.

הריצו את התא הבא. מה הודפס, ומה לא?

</div>

```python
# This is a comment line
print("רק השורה הזו תודפס")
```

<div dir="rtl">

### הערת מקטע: `'''`

אם רוצים לכתוב הערה של כמה שורות, תוחמים אותה בשלושה גרשיים בהתחלה ושלושה גרשיים בסוף.

ההערות מופיעות רק בקוד, ולא מודפסות על המסך.

</div>

```python
'''
More than
one comment line
:-)
'''
print("ההערה למעלה לא הודפסה")
```
